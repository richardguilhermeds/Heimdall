```python
import os
import pandas as pd
from pathlib import Path

# Define paths
raw_path = Path(r"G:\Meu Drive\Heimdall\data\raw\resumo")
processed_path = Path(r"G:\Meu Drive\Heimdall\data\processed\resumo")

# Create processed directory if it doesn't exist
processed_path.mkdir(parents=True, exist_ok=True)

dfs = []
for file in raw_path.glob("conglomerados_prudenciais*.csv"):
    try:
        df = pd.read_csv(file, on_bad_lines='skip', sep=';')
        df['Instituição'] = df['Instituição'].str.replace(r'\s*-\s*PRUDENCIAL', '', regex=True)
        if 'Código' in df.columns:
            df.drop('Código', axis=1, inplace=True)
        dfs.append(df)
        print(f"Carregando: {file.name}")
    except Exception as e:
        print(f"Erro ao carregar {file.name}: {e}")

# Concatenate all dataframes
if dfs:
    df_consolidated = pd.concat(dfs, ignore_index=True)
    out_file = processed_path / "conglomerados_prudenciais_resumo_consolidado.csv"
    df_consolidated.to_csv(out_file, index=False)
    print("Todos os arquivos foram consolidados com sucesso!")
    print(f"Total de linhas: {len(df_consolidated)}")
else:
    print("Nenhum arquivo foi carregado com sucesso.")
```
